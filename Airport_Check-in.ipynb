```python
class PassengerNode:
    def __init__(self, name):
        self.name = name
        self.next = None


class AirportQueue:
    def __init__(self):
        self.front = None
        self.rear = None

    # Add passenger
    def enqueue(self, name):
        new_passenger = PassengerNode(name)

        if self.rear is None:
            self.front = self.rear = new_passenger
        else:
            self.rear.next = new_passenger
            self.rear = new_passenger

        print("Passenger added:", name)

    # Check-in passenger
    def dequeue(self):
        if self.front is None:
            print("No passengers in queue")
        else:
            print("Passenger checked-in:", self.front.name)
            self.front = self.front.next

            if self.front is None:
                self.rear = None

    # View first passenger
    def peek(self):
        if self.front is None:
            print("No passengers in queue")
        else:
            print("Next passenger:", self.front.name)

    # Display passengers
    def display(self):
        if self.front is None:
            print("No passengers in queue")
        else:
            temp = self.front

            print("Passengers in queue:")

            while temp is not None:
                print(temp.name)
                temp = temp.next

    # Count passengers
    def count(self):
        count = 0
        temp = self.front

        while temp is not None:
            count += 1
            temp = temp.next

        print("Total passengers:", count)


queue = AirportQueue()

n = int(input("Enter number of passengers: "))

for i in range(n):
    name = input("Enter passenger name: ")
    queue.enqueue(name)

print()

queue.display()

queue.count()

queue.peek()

print()

queue.dequeue()

print()

queue.display()
queue.count()
queue.peek()
```
